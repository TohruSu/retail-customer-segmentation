This independent portfolio exercise starts from fictional records generated with a fixed random seed. It uses no original survey questions, records, distributions, or analytical outputs. The dataset is a coding fixture, not evidence about customers.

See [the generator assumptions](../docs/SYNTHETIC_DATA.md) and [publication boundaries](../docs/PROVENANCE_AND_SCOPE.md).

```python
import pandas as pd
from IPython.display import display

from retail_case.schema import ATTITUDE_COLUMNS, BEHAVIOR_COLUMNS, SYNTHETIC_NOTICE
from retail_case.synthetic import generate_data

df = generate_data(n=720, seed=2026, missing_rate=0.03)


def show_synthetic_table(title, table):
    print(f"SYNTHETIC — {title}")
    print(SYNTHETIC_NOTICE)
    display(table)
```

## Schema and validation

Synthetic identifiers identify fictional rows only. Store visits and other-channel visits are behavioral counts; seven invented attitudes are bounded proxies. Clustering will use attitudes only, excluding identifiers and both behavior fields. Controlled missingness is retained here and handled explicitly during preprocessing.

```python
from retail_case.validation import validate_data

validation_summary = validate_data(df)
numeric_columns = list(BEHAVIOR_COLUMNS) + list(ATTITUDE_COLUMNS)
schema_summary = pd.DataFrame(
    {
        "dtype": df[numeric_columns].dtypes.astype(str),
        "observed_min": df[numeric_columns].min(),
        "observed_max": df[numeric_columns].max(),
        "missing_rows": df[numeric_columns].isna().sum(),
        "missing_share": df[numeric_columns].isna().mean(),
    }
).rename_axis("invented_feature")
show_synthetic_table("Observed schema, ranges, and missingness", schema_summary)
show_synthetic_table("Validation report", pd.DataFrame({"validation_result": [validation_summary]}))
```

## What this establishes

**Observation:** the checks above evaluate this generated fixture's columns, IDs, types, bounds, finite values, and missingness. **Assumption:** the generator's ranges and noise are suitable for exercising an analytics workflow. **Unknown:** no measurement-validity or real population claim follows from passing validation.
