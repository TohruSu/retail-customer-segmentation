# 06 · A qualitative 3C case and channel comparison

**Evidence status: HYPOTHETICAL CASE.** This stage uses generic retail channels and editorial assumptions. It makes no factual claim about a named company's capabilities or a measured customer's needs. No survey-derived competitor conclusion is used. See [the strategy case](../docs/STRATEGY_CASE.md).

The preceding [targeting stage](05_targeting.ipynb) demonstrated a conditional reasoning method on synthetic data. It did not establish an actual target audience. This notebook runs independently and creates only qualitative tables.

```python
import pandas as pd
from IPython.display import display


def show_case_table(title, rows):
    print(f"HYPOTHETICAL CASE — {title}; qualitative assumptions, not measured findings.")
    display(pd.DataFrame(rows))
```

```python
show_case_table(
    "Customer / company / competitor evidence boundary",
    [
        {
            "lens": "Customer",
            "status": "ASSUMPTION",
            "case_statement": (
                "Some meal occasions may require balancing time, cost, effort, and variety."
            ),
            "evidence_needed": "Permitted task interviews and observed comparison difficulties",
        },
        {
            "lens": "Company / host",
            "status": "ASSUMPTION",
            "case_statement": (
                "A hypothetical host has an existing food floor and can maintain changing meal "
                "themes."
            ),
            "evidence_needed": (
                "Actual operating capacity, assortment, signage rights, and staffing"
            ),
        },
        {
            "lens": "Competitor / substitute",
            "status": "INTERPRETATION",
            "case_statement": "Different channels may complete parts of the same meal occasion.",
            "evidence_needed": "Local channel audit and real customer route research",
        },
        {
            "lens": "Decision",
            "status": "RECOMMENDATION",
            "case_statement": (
                "Test whether cross-option comparison adds value beyond existing shopping paths."
            ),
            "evidence_needed": "Usability tests and a comparative pilot",
        },
    ],
)
```
