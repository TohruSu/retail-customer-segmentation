# 05 · Turn descriptive profiles into conditional hypotheses

This stage separates measured descriptions of the fictional fixture from editorial decisions. **All numerical tables are SYNTHETIC:** illustrative results on independently generated synthetic data; not findings from the ADK consumer survey.

This notebook independently regenerates the fixture and evaluates the analytical pipeline. Cluster IDs are arbitrary local model labels. They are not original survey labels, personas, stable population identities, or a ranking of valuable customers.

```python
import pandas as pd
from IPython.display import display

from retail_case.schema import ATTITUDE_COLUMNS, SYNTHETIC_NOTICE
from retail_case.synthetic import generate_data

df = generate_data(n=720, seed=2026, missing_rate=0.03)


def show_synthetic_table(title, table):
    print(f"SYNTHETIC — {title}")
    print(SYNTHETIC_NOTICE)
    display(table)
```

```python
from retail_case.clustering import analyze_data

result = analyze_data(df)
profiles = pd.DataFrame(result["profiles"]).set_index("cluster")
show_synthetic_table(
    "Exploratory partition descriptions; raw observed means omit missing cells", profiles
)
baseline = result["baseline"]["unsegmented"]
show_synthetic_table("Unsegmented descriptive baseline", pd.DataFrame([baseline]))
show_synthetic_table(
    "Transparent descriptive strata baseline",
    pd.DataFrame(result["baseline"]["descriptive_strata"]),
)
print(f"SYNTHETIC descriptive stratification rule: {result['baseline']['stratification_rule']}")
```

```python
attitude_means = profiles[list(ATTITUDE_COLUMNS)]
baseline_means = pd.Series({column: baseline[column] for column in ATTITUDE_COLUMNS})
deviations = attitude_means.subtract(baseline_means, axis="columns")
show_synthetic_table(
    "Profile mean differences from the fixture-wide baseline; descriptive only", deviations
)

hypotheses = pd.DataFrame(
    {
        "conditional_time_pressure_task": (deviations["shopping_time_pressure"] > 0)
        & (deviations["convenience_preference"] > 0),
        "conditional_discovery_task": (deviations["meal_planning_interest"] > 0)
        & (deviations["discovery_interest"] > 0),
    }
)
show_synthetic_table(
    "Illustrative editorial conditions, not a target score or recommendation rank", hypotheses
)
```

## Decision boundary

| Status | Statement | What would change the decision? |
| --- | --- | --- |
| Observation | Model profiles differ within the generated fixture; differences have no population interpretation. | Independent, permitted measurements of actual tasks and needs. |
| Assumption | A comparison experience might help people balancing time, effort, cost, and meal variety. | Task research showing what choices are difficult and when. |
| Interpretation | Above-baseline fictional means illustrate how a conditional audience hypothesis can be written. | Replication, uncertainty estimates, and usefulness in an actual setting. |
| Recommendation | Compare a broad experience with a task-focused variant only after real task evidence supports it. | Pilot usability, access, operational fit, and valid comparative outcomes. |
| Unknown | Which audience, if any, benefits or produces incremental purchases or revisits. | A designed real-world pilot; these synthetic records cannot answer it. |

No synthetic share becomes a target-market size, customer value, or impact forecast. Keep the no-segmentation option. [06 · Competitor / 3C](06_competitor_3c.ipynb) carries the conditional need into a qualitative case discussion without converting it into a consumer claim.
